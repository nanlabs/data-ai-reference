### **4. Installing and Using DuckDB for Geospatial Queries**

To improve performance and run SQL-based geospatial queries, DuckDB is installed.

```python
pip install duckdb -q
```

```python
import duckdb

conn = duckdb.connect()
```

Enabling the DuckDB Spatial Extension
DuckDB requires spatial extensions to work with geospatial data.

```python
conn.query('INSTALL spatial')
conn.query('LOAD spatial')
```

Defining a Spatial SQL Query
A SQL query is defined to process geospatial data directly from a CSV file.

📌 What Happens Here?

* read_csv(): Reads a CSV file directly from an S3 bucket.
* ST_Point(column09, column08): Converts latitude (column08) and longitude (column09) into a POINT geometry.
* ST_AsText(): Returns the WKT (Well-Known Text) representation of the geometry.

```python
# Define a DuckDB SQL query
# Replace "your_table" with the table or query containing lat/lon or WKT geometry

duckdb_query = """
SELECT *, ST_AsText(ST_Point(column09, column08)) as geom
FROM read_csv('https://s3.amazonaws.com/tripdata/202406-citibike-tripdata.zip', ignore_errors=true)
"""
```

Converting Query Results to Pandas DataFrame
After running the SQL query, the results are converted to Pandas.

```python
duckdb_df = conn.execute(duckdb_query).fetchdf()
```

```python
duckdb_df.head(10)
```

```python
duckdb_gdf = gpd.GeoDataFrame(duckdb_df, geometry=gpd.GeoSeries.from_wkt(duckdb_df['geom']), crs='EPSG:4326')
```

```python
duckdb_gdf.info()
```

```python
duckdb_gdf.head(100).explore()
```

```python
nbh_query = """
SELECT 
    count(b.column00) as count,
    n.neighborhood,
    ST_AsText(n.geom) as geom
FROM 
    ST_Read('https://raw.githubusercontent.com/HodgesWardElliott/custom-nyc-neighborhoods/refs/heads/master/custom-pedia-cities-nyc-Mar2018.geojson') n
JOIN 
    read_csv('https://s3.amazonaws.com/tripdata/202406-citibike-tripdata.zip', ignore_errors=true) b
ON 
    ST_Intersects(n.geom, ST_Point(column09, column08))
WHERE 
    CAST(column02 AS DATE) = DATE '2024-06-15'
GROUP BY 
    n.neighborhood, n.geom
"""
```

```python
duckdb_df = conn.query(nbh_query).to_df()
duckdb_df.head(10)
```

```python
join_gdf = gpd.GeoDataFrame(duckdb_df, geometry=gpd.GeoSeries.from_wkt(duckdb_df['geom']), crs='EPSG:4326')
```
